```python
from openai import OpenAI
import json

client = OpenAI(
    base_url="http://localhost:8000",
    api_key="api_key",
)
```

```python
# clear web_retrieval store
client.ops.web_retrieval.delete()
# populate web_retrieval store
crawl = client.ops.web_retrieval.crawl_and_upsert(
    root_urls=["https://www.spacex.com/"],
    max_depth=1,
    description="Recent news from spacex.",
    constrain_to_root_domain=False,
)
```

```python
# upload the file 
file = client.files.create(
    purpose="assistants",
    file=open("../assistants_api/assets/my_information.txt", "rb"),
)
# vectorize, split, then store the file
vector_store = client.beta.vector_stores.create(
    name="Information About Me",
    file_ids=[file.id],
)
```

```python
# create assistant
assistant = client.beta.assistants.create(
    instructions="""YOU MUST USE TOOLS! You empersonate Daniel based on the information in the file my_information.txt which you can obtain using the file_search tool.""",
    tools=[
        {"type": "web_retrieval"},
        {"type": "file_search"}
    ],
    tool_resources={
        "file_search": {
            "vector_store_ids": [vector_store.id]
        }
    },
    model="phi3:medium"
)
# create thread
thread = client.beta.threads.create(
    messages=[
        {
            "role": "user",
            "content": "Concisely tell me about Daniel",  # noqa
        },
    ],
)
# run response generation
run = client.beta.threads.runs.create_and_poll(
    thread_id=thread.id,
    assistant_id=assistant.id,
)
# print messages
messages = client.beta.threads.messages.list(
    thread_id=thread.id, order='desc'
)
print(json.dumps(messages.model_dump(), indent=2))
```
